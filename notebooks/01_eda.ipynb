```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/indicators_mdg.csv")
df.head(3)
```

```python
df = df.rename(columns={'Country Name':'country_name', 'Country ISO3':'country_iso', 'Year':'year', 'Indicator Name':'indicator_name', 'Indicator Code':'indicator_code', 'Value':'value'})
df.sample(5)
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df['indicator_name'].describe()
```

```python
df['year'].describe().astype(int)
```

```python
df['value'] = df['value'].apply(
    lambda x: np.float64(x).round(2)
)
df.sample(10)
```

```python
df.groupby('indicator_name')['value'].describe()
```

```python
df['theme_code'] = df['indicator_code'].apply(
    lambda x: str(x).split('.')[0].split('_')[0].upper()
)
df.sample(5)
```

```python
df.groupby('theme_code')['value'].describe()
```
